```python
# Bài 1
class findNumber:
    def number(self):
        Numbers = []
        for number in range(2000, 3201):
            if number % 7 == 0 and number % 5 != 0:
                Numbers.append(str(number))
        return Numbers
finder = findNumber()       
result = finder.number()
print(",".join(result))   
```

```python
# Bài 2
class giaithua:
    def tinhgiaithua(self):
        n = int(input("Nhập số nguyên dương n: "))
        if n < 0:
            return "Không tồn tại giai thừa của số âm"
        elif n == 0:
            return 1
        else:
            giai_thua = 1
            for i in range(1, n + 1):
                giai_thua *= i
            return giai_thua
pheptinh = giaithua()
ketqua = pheptinh.tinhgiaithua()
print(f"Giai thừa của {n} là {ketqua}")
```

```python
# Bài 3
class dictionary:
    def dictionaryOfnumbers(self):
        dicNumbers = {}
        n = int(input("Nhập số nguyên dương n: "))
        for i in range(1, n + 1):
            dicNumbers[i] = i * i
        return dicNumbers
pheptinh = dictionary()
ketqua = pheptinh.dictionaryOfnumbers()
print(ketqua)
```

```python
# Bài 4
class listAndTuple:
    def listAndTupleOfnumbers(self):
        numbers = input("Nhập dãy số, cách nhau bằng dấu phẩy: ")
        listNumbers = numbers.split(",")
        tupleNumbers = tuple(listNumbers)
        return listNumbers, tupleNumbers
pheptinh = listAndTuple()
ketqua = pheptinh.listAndTupleOfnumbers()
print("Danh sách:", ketqua[0])
print("Tuple:", ketqua[1])
```

```python
# Bài 5 
class StringUpper:
    def __init__(self):
        self.text = ""   
    def getString(self):
        self.text = input("Nhập chuỗi: ")
    def printString(self):
        return self.text.upper()
converter = StringUpper()
converter.getString()
ket_qua = converter.printString()
print("Chuỗi sau khi in hoa:", ket_qua)
if ket_qua.isupper():
    print("Kiểm tra: chuỗi đã được in hoa.")
else:
    print("Kiểm tra: chuỗi chưa được in hoa.")
```

```python
# Bài 9
import datetime
class whatIsYourAge:
    def tinhTuoi(self):
        inputDateTime = datetime.datetime.strptime(input("Nhập ngày sinh của bạn (YYYY-MM-DD): "), "%Y-%m-%d").date()
        today = datetime.datetime.now().date()
        Age = (today - inputDateTime).days // 365
        return Age
pheptinh = whatIsYourAge()
ketqua = pheptinh.tinhTuoi()
print(f"Tuổi của bạn là: {ketqua}")
```

```python
# Bài 10
class payLoad:
    def calculatePayLoad(self):
        workHours = int(input("Nhập số giờ làm việc: "))
        payRate = int(input("Nhập mức lương theo giờ: "))
        if workHours <= 44:
            salary =  workHours * payRate
        else:
            salary = 44 * payRate + (workHours - 44) * payRate * 1.5
        return salary
pheptinh = payLoad()
ketqua = pheptinh.calculatePayLoad()
print(f"Tổng lương của bạn là: {ketqua}")
```

```python
# Bài 11
import math
class calSqrt:
    def calculateSqrt(self):
        C = 50
        H = 30
        D = input("Nhập các giá trị D, cách nhau bằng dấu phẩy: ").split(",")
        Q = []
        for d in D:
            q = round(math.sqrt((2 * C * int(d)) / H))
            Q.append(q)
        return Q
pheptinh = calSqrt()
ketqua = pheptinh.calculateSqrt()
print(f"Giá trị Q tương ứng với các giá trị D đã nhập là: {ketqua}")
```

```python
# Bài 12
class twoDimensionArray:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def dimension(self):
        dimensionArr = []
        for i in range(self.x):
            dimensionArrChild = []
            for j in range(self.y):
                dimensionArrChild.append(i * j)
            dimensionArr.append(dimensionArrChild)
        return dimensionArr
X, Y = map(int, input("Nhập X, Y (cách nhau bằng dấu phẩy): ").split(","))
matrix = twoDimensionArray(X, Y)
print(matrix.dimension())
```

```python
# Bài 13
class alphabetSort:
    def __init__(self, input_string):
        self.input_string = input_string
    def sortwords(self):
        words = []
        for word in self.input_string.split(","):
            word = word.strip()
            if word:
                words.append(word)
        words.sort()
        return ",".join(words)
input_string = input("Nhập các từ, cách nhau bằng dấu phẩy: ")
sorter = alphabetSort(input_string)
print(sorter.sortwords())
```

```python
# Bài 14
class upperCase:
    def __init__(self, input_string):
        self.input_string = input_string
    def convertToUpperCase(self):
        return self.input_string.upper()
input_string = input("Nhập một chuỗi: ")
upper_caser = upperCase(input_string)
print(upper_caser.convertToUpperCase())
```

```python
# Bài 15
class removeDupAndSort:
    def __init__(self, input_string):
        self.input_string = input_string
    def removeDuplicates(self):
        words = self.input_string.split()
        return set(words)
    def sortWord(self):
        uniqueWord = self.removeDuplicates()
        return " ".join(sorted(uniqueWord))
input_string = input("Nhập các từ, cách nhau bằng khoảng trắng: ")
processor = removeDupAndSort(input_string)
print(processor.sortWord())
```
