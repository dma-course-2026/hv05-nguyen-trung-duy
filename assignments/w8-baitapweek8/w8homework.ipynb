```python
# Bài 1
class findNumber:
    def number(self):
        Numbers = []
        for number in range(2000, 3201):
            if number % 7 == 0 and number % 5 != 0:
                Numbers.append(str(number))
        return Numbers
finder = findNumber()       
result = finder.number()
print(",".join(result))   
```

```python
# Bài 2
class giaithua:
    def tinhgiaithua(self):
        n = int(input("Nhập số nguyên dương n: "))
        if n < 0:
            return "Không tồn tại giai thừa của số âm"
        elif n == 0:
            return 1
        else:
            giai_thua = 1
            for i in range(1, n + 1):
                giai_thua *= i
            return giai_thua
pheptinh = giaithua()
ketqua = pheptinh.tinhgiaithua()
print(f"Giai thừa của {n} là {ketqua}")
```

```python
# Bài 3
class dictionary:
    def dictionaryOfnumbers(self):
        dicNumbers = {}
        n = int(input("Nhập số nguyên dương n: "))
        for i in range(1, n + 1):
            dicNumbers[i] = i * i
        return dicNumbers
pheptinh = dictionary()
ketqua = pheptinh.dictionaryOfnumbers()
print(ketqua)
```

```python
# Bài 4
class listAndTuple:
    def listAndTupleOfnumbers(self):
        numbers = input("Nhập dãy số, cách nhau bằng dấu phẩy: ")
        listNumbers = numbers.split(",")
        tupleNumbers = tuple(listNumbers)
        return listNumbers, tupleNumbers
pheptinh = listAndTuple()
ketqua = pheptinh.listAndTupleOfnumbers()
print("Danh sách:", ketqua[0])
print("Tuple:", ketqua[1])
```

```python
# Bài 5 
class StringUpper:
    def __init__(self):
        self.text = ""   
    def getString(self):
        self.text = input("Nhập chuỗi: ")
    def printString(self):
        return self.text.upper()
converter = StringUpper()
converter.getString()
ket_qua = converter.printString()
print("Chuỗi sau khi in hoa:", ket_qua)
if ket_qua.isupper():
    print("Kiểm tra: chuỗi đã được in hoa.")
else:
    print("Kiểm tra: chuỗi chưa được in hoa.")
```

```python
# Bai 6
class binhphuong:
    def tinhbinhphuong(self):
        global n
        n = int(input("Nhập số nguyên dương n: "))
        binh_phuong = n ** 2
        return binh_phuong
phep_tinh = binhphuong()
ket_qua = phep_tinh.tinhbinhphuong()
print(f"Bình phương của {n} là {ket_qua}")
```
