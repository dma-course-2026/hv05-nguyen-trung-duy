```python
# Bài 50 - Viết chương trình tìm tất cả các số có 3 chữ số sao cho tổng tất cả các chữ số bằng tích của chúng
def hundreds_number():
    result = []
    for i in range(100, 1000):
        hundreds = i // 100
        tens = (i // 10) % 10
        units = i % 10
        if hundreds + tens + units == hundreds * tens * units:
            result.append(i)
    return result
print(hundreds_number())
```

```python
# Bài 52 - Tìm tất cả các tổ hợp có được của 3 loại tiền trên cho số tiền vừa nhập
def find_combinations():
    amount_exchange = int(input("Nhập số tiền cần đổi: "))
    denominations1 = int(input("Nhập mệnh giá loại tiền thứ nhất: "))
    denominations2 = int(input("Nhập mệnh giá loại tiền thứ hai: "))
    denominations3 = int(input("Nhập mệnh giá loại tiền thứ ba: "))
    combinations = []
    for i in range(0, amount_exchange // denominations1 + 1):  
        for j in range(0, amount_exchange // denominations2 + 1): 
            for k in range(0, amount_exchange // denominations3 + 1):  
                if (i * denominations1 + j * denominations2 + k * denominations3) == amount_exchange:
                    combinations.append((i, j, k))
    return combinations
print(find_combinations())
```

```python
# Bài 57 - Tính C(k, n)
def tohop():
    n = int(input("Nhập vào số n: "))
    k = int(input("Nhập vào số k: "))
    gt_n = 1
    gt_k = 1
    gt_n_k = 1
    for i in range(1, n + 1):
        gt_n *= i
    for i in range(1, k + 1):
        gt_k *= i
    for i in range(1, n - k + 1):
        gt_n_k *= i
    C_k_n = gt_n // (gt_k * gt_n_k)
    print(f"C({k}, {n}) = {C_k_n}")
    return C_k_n
tohop()
```

```python
# Bai 58 - chương trình tìm 10 số Fibonacci đầu tiên
def fibonacci():
    n = 10
    fibArr = []
    a, b = 0, 1
    for i in range(n):
        fibArr.append(a)
        a, b = b, a + b
    print(f"{n} số Fibonacci đầu tiên là: {fibArr}")
    return fibArr
fibonacci()
```

```python
# Bài 59 - Viết chương trình tìm ước số chung lớn nhất (USCLN) và bội số chung nhỏ nhất (BSCNN) của 2 số nguyên dương a và b nhập từ bàn phím.
def uscln_bscnn():
    a = int(input("Nhập số nguyên dương a: "))
    b = int(input("Nhập số nguyên dương b: "))
    uscln = 1
    for i in range(1, min(a, b) + 1):
        if a % i == 0 and b % i == 0:
            uscln = i
    bscnn = a * b // uscln
    print(f"USCLN của {a} và {b} là: {uscln}")
    print(f"BSCNN của {a} và {b} là: {bscnn}")
uscln_bscnn()
```

```python
# Bài 60 - Viết chương trình liệt kê n số nguyên tố đầu tiên trong Python. Số nguyên dương n được nhập từ bàn phím
def is_prime():
    n = int(input("Nhập số nguyên dương n: "))
    primes = []
    for num in range(2, n * n + 2):
        so_uoc = 0
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                so_uoc += 1
        if so_uoc == 0 and len(primes) < n:
            primes.append(num)
    print(f"{n} số nguyên tố đầu tiên là: {primes}")
    return primes
is_prime()
```

```python
# Bài 63 - Viết chương trình phân tích số nguyên n thành các thừa số nguyên tố trong Python.
def phan_tich_thua_so_nguyen_to():
    n = int(input("Nhập số nguyên dương n: "))
    thua_so_nguyen_to = []
    for i in range(2, n + 1):
        for _ in range(n):
            if n % i == 0:
                thua_so_nguyen_to.append(i)
                n //= i
    print(f"Các thừa số nguyên tố là: {thua_so_nguyen_to}")
    return thua_so_nguyen_to
phan_tich_thua_so_nguyen_to()
```

```python
# Bài 81 - Viết chương trình nhập n số nguyên từ bàn phím. Hãy cho biết dãy số vừa nhập có bao nhiêu số nguyên tố và thông báo kết quả ra màn hình
def dem_so_nguyen_to():
    n = int(input("Nhập số lượng phần tử: "))
    arr = []
    for i in range(n):
        phan_tu = int(input(f"Nhập phần tử thứ {i + 1}: "))
        arr.append(phan_tu)
    so_nguyen_to = 0
    for phan_tu in arr:
        if phan_tu > 1:
            is_prime = True
            for i in range(2, int(phan_tu ** 0.5) + 1):
                if phan_tu % i == 0:
                    is_prime = False
                    break
            if is_prime:
                so_nguyen_to += 1
    print(f"Số lượng số nguyên tố trong dãy là: {so_nguyen_to}")
    return so_nguyen_to
dem_so_nguyen_to()
```

```python
# Bài 82 - Viết chương trình nhập n số nguyên từ bàn phím. Tính tổng các số nguyên tố trong dãy và thông báo kết quả ra màn hình
def sumSoNguyenTo():
    n = int(input("Nhập số lượng phần tử: "))
    arr = []
    for i in range(n):
        phan_tu = int(input(f"Nhập phần tử thứ {i + 1}: "))
        arr.append(phan_tu)
    tong_nguyen_to = 0
    for phan_tu in arr:
        if phan_tu > 1:
            is_prime = True
            for i in range(2, int(phan_tu ** 0.5) + 1):
                if phan_tu % i == 0:
                    is_prime = False
                    break
            if is_prime:
                tong_nguyen_to += phan_tu
    print(f"Tổng các số nguyên tố trong dãy là: {tong_nguyen_to}")
    return tong_nguyen_to
sumSoNguyenTo()
```

```python
# Bài 83 - Cho dãy A gồm n phần tử có giá trị ngẫu nhiên từ -300 đến 300. Viết chương trình tạo dãy B[1..n], trong đó B[i] là tổng giá trị của i phần tử đầu tiên của dãy A.
import random
def create_array_B():
    n = int(input("Nhập số lượng phần tử n: "))
    A = [random.randint(-300, 300) for _ in range(n)]
    B = []
    current_sum = 0
    for i in range(n):
        current_sum += A[i]
        B.append(current_sum)
    print(f"Dãy A: {A}")
    print(f"Dãy B: {B}")
    return A, B
create_array_B()
```

```python
# Bài 84 - Cho dãy A gồm n phần tử có giá trị ngẫu nhiên từ -100 đến 100. Viết chương trình kiểm tra xem dãy A có lập thành cấp số cộng hay không và thông báo kết quả ra màn hình.
import random
def is_arithmetic_sequence():
    n = int(input("Nhập số lượng phần tử n: "))
    A = [random.randint(-100, 100) for _ in range(n)]
    if n < 2:
        print("Dãy có ít hơn 2 phần tử, không thể xác định cấp số cộng.")
        return False
    common_difference = A[1] - A[0]
    is_arithmetic = all(A[i] - A[i - 1] == common_difference for i in range(2, n))
    print(f"Dãy A: {A}")
    if is_arithmetic:
        print("Dãy A lập thành cấp số cộng.")
    else:
        print("Dãy A không lập thành cấp số cộng.")
    return is_arithmetic
is_arithmetic_sequence()
```

```python
# Bài 85 - Cho dãy A gồm n phần tử có giá trị ngẫu nhiên từ -100 đến 100. Hãy đếm số lượng số chẵn, số lẻ trong dãy và in kết quả ra màn hình.
import random
def countchanle():
    n = int(input("Nhập số lượng phần tử n: "))
    A = [random.randint(-100, 100) for _ in range(n)]
    count_even = sum(1 for x in A if x % 2 == 0)
    count_odd = n - count_even
    print(f"Dãy A: {A}")
    print(f"Số lượng số chẵn: {count_even}")
    print(f"Số lượng số lẻ: {count_odd}")
    return count_even, count_odd
countchanle()
```

```python
# Bài 88 - Cho số tự nhiên(1000<N<10⁶). Viết chương trình tìm chữ số lớn nhất của số N.
def find_max_digit():
    N = int(input("Nhập số tự nhiên N (1000 < N < 10^6): "))
    if 1000 < N < 10**6:
        max_digit = max(int(digit) for digit in str(N))
        print(f"Chữ số lớn nhất của số {N} là: {max_digit}")
        return max_digit
    else:
        print("Số nhập vào không hợp lệ. Vui lòng nhập lại.")
        return None
find_max_digit()
```

```python
# Bài 89 - Nhập vào hai xâu từ bàn phím, in ra màn hình xâu dài hơn
def longer_string():
    string1 = input("Nhập xâu thứ nhất: ")
    string2 = input("Nhập xâu thứ hai: ")
    if len(string1) > len(string2):
        print(f"Xâu dài hơn là: {string1}")
        return string1
    elif len(string2) > len(string1):
        print(f"Xâu dài hơn là: {string2}")
        return string2
    else:
        print("Hai xâu có độ dài bằng nhau.")
        return None
longer_string()
```

```python
# Bài 90 - Nhập vào một xâu, in ra màn hình xâu đảo ngược
def reverse_string():
    input_string = input("Nhập vào một xâu: ")
    reversed_string = input_string[::-1]
    print(f"Xâu đảo ngược là: {reversed_string}")
    return reversed_string
reverse_string()
```

```python
# Bài 91 - Nhập vào một xâu, kiểm tra xâu đó có phải xâu đối xứng hay không?
def xauDoixung():
    input_string = input("Nhập vào một xâu: ")
    if input_string == input_string[::-1]:
        print(f"Xâu '{input_string}' là xâu đối xứng.")
        return True
    else:
        print(f"Xâu '{input_string}' không phải là xâu đối xứng.")
        return False
xauDoixung()
```

```python
# Bài 92 - Nhập vào một xâu, thay tất cả cụm từ 'anh' bằng cụm từ 'em'
def replace_anh_with_em():
    input_string = input("Nhập vào một xâu: ")
    replaced_string = input_string.replace('anh', 'em')
    print(f"Xâu sau khi thay thế: {replaced_string}")
    return replaced_string
replace_anh_with_em()
```

```python
# Bai 93 - Nhập vào một xâu s từ bàn phím. Hãy cho biết xâu s có bao nhiêu kí tự là chữ cái tiếng anh(không phân biệt chữ thương và chữ hoa)
def count_english_letters():
    input_string = input("Nhập vào một xâu: ")
    count = sum(1 for char in input_string if char.isalpha())
    print(f"Số lượng kí tự là chữ cái tiếng Anh trong xâu là: {count}")
    return count
count_english_letters()
```

```python
# Bài 94 - Nhập vào một xâu s từ bàn phím và một kí tự x bất kì. Hãy cho biết kí tự x xuất hiện trong xâu s bao nhiêu lần và thông báo kết quả ra màn hình
def countchar():
    input_string = input("Nhập vào một xâu: ")
    char_to_count = input("Nhập kí tự cần đếm: ")
    count = input_string.count(char_to_count)
    print(f"Kí tự '{char_to_count}' xuất hiện {count} lần trong xâu.")
    return count
countchar()
```

```python
# Bài 95 - Nhập vào một xâu s từ bàn phím, tạo một xâu mới gồm các chữ số có trong xâu s và in kết quả ra màn hình
def extract_digits():
    input_string = input("Nhập vào một xâu: ")
    digits_string = ''.join(filter(str.isdigit, input_string))
    print(f"Các chữ số có trong xâu là: {digits_string}")
    return digits_string
extract_digits()
```

```python
# Bài 96 - Nhập vào một xâu s từ bàn phím, tạo xâu s2 từ xâu s bằng cách loại bỏ các dấu cách có trong xâu s.
def remove_spaces():
    input_string = input("Nhập vào một xâu: ")
    string_without_spaces = input_string.replace(' ', '')
    print(f"Xâu sau khi loại bỏ dấu cách: {string_without_spaces}")
    return string_without_spaces
remove_spaces()
```

```python
# Bài 97 - Nhập xâu họ tên s, hãy chuẩn hóa lại xâu họ tên đó. Xâu họ tên là xâu chuẩn hóa khi thỏa mãn các điều sau:
# Không có dấu cách thừa
# Chữ cái đầu mỗi từ viết in hoa còn lại là in thường.
def name():
    input_string = input("Nhập vào họ tên: ")
    # Loại bỏ dấu cách thừa và chuẩn hóa chữ cái
    normalized_name = ' '.join(word.capitalize() for word in input_string.split())
    print(f"Họ tên sau khi chuẩn hóa: {normalized_name}")
    return normalized_name
name()
```

```python
# Bài 98 - Nhập vào hai xâu s1, s2 từ bàn phím, cho biết xâu s2 xuất hiện trong xâu s1 bao nhiêu lần và in kết quả ra màn hình
def countString():
    s1 = input("Nhập xâu s1: ")
    s2 = input("Nhập xâu s2: ")
    count = s1.count(s2)
    print(f"Xâu '{s2}' xuất hiện {count} lần trong xâu '{s1}'.")
    return count
countString()
```

```python
# Bài 99 - Nhập vào một xâu từ bàn phím, loại bỏ các kí tự trùng lặp trong xâu và in kết quả ra màn hình.
def delDup():
    input_string = input("Nhập vào một xâu: ")
    result_string = ''.join(sorted(set(input_string), key=input_string.index))
    print(f"Xâu sau khi loại bỏ các kí tự trùng lặp: {result_string}")
    return result_string
delDup()
```

```python
# Bài 100 - Nhập vào một xâu từ bàn phím, tách các số có trong xâu và in kết quả ra màn hình.
def delNum():
    input_string = input("Nhập vào một xâu: ")
    result_string = ''.join(filter(lambda x: not x.isdigit(), input_string))
    print(f"Xâu sau khi loại bỏ các số: {result_string}")
    return result_string
delNum()
```
