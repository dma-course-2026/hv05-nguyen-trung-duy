```python
# Bai 5 - Diem thi dua
def diemthidua():
    # Nhap truong hop bi tru diem thi dua
    t, n, m = map(int, input("Nhap so truong hop bi tru diem thi dua: ").split())
    # Nhap so diem thi dua bi tru
    a, b, c = map(int, input("Nhap so diem thi dua bi tru theo tung truong hop: ").split())
    # Tinh diem bi tru
    tongdiembitru = t * a + n * b + m * c
    return tongdiembitru
diemthidua()
```

```python
# Bai 6 - Dien tich khu vuon
def dientichconlaicuakhuvuon():
    # Nhap chieu dai va chieu rong
    Chieudaib = float(input("Nhap chieu dai khu vuon: "))
    Chieuronga = float(input("Nhap chieu rong khu vuon: "))
    # Tinh dien tich khu vuon
    dientichcuakhuvuon = Chieudaib * Chieuronga
    # Tinh dien tich khu vui choi
    dientichkhuvuichoi = 3.14 * (Chieudaib / 2) ** 2
    # Tinh dien tich con lai cua khu vuon
    dientichconlaicuakhuvuon = dientichcuakhuvuon - dientichkhuvuichoi
    # In ra ket qua
    print("Dien tich con lai cua khu vuon la:", dientichconlaicuakhuvuon)
    return dientichconlaicuakhuvuon
dientichconlaicuakhuvuon()
```

```python
# Bai 7 - Diem trung binh
def diemtrungbinh():
    # Nhap diem
    diem1 = float(input("Nhap diem mon hoc thu nhat: "))
    diem2 = float(input("Nhap diem mon hoc thu hai: "))
    diem3 = float(input("Nhap diem mon hoc thu ba: "))
    # Tinh diem trung binh
    diemtrungbinh = (diem1 + diem2 + diem3) / 3
    # In ra ket qua diem trung binh
    print("Diem trung binh cua 3 mon hoc la:", f"{diemtrungbinh:.4f}")
    # Xet xem Dat hay Khong Dat
    if diemtrungbinh >= 5:
        print("Qua mon")
    else:
        print("Khong qua mon")
    #Tra ket qua
    return diemtrungbinh
diemtrungbinh()
```

```python
# Bai 9 - Tinh tien dien
def tinhtiendien():
    hovatenchuho = input("Nhap ho va ten chu ho: ")
    chiso1 = float(input("Nhap chi so dien cu: "))
    chiso2 = float(input("Nhap chi so dien moi: "))
    #tinh tien dien phai tra
    if chiso2 - chiso1 <= 60:
        tienphaitra = (chiso2 - chiso1) * 5
    elif 61 <= chiso2 - chiso1 <= 160:
        tienphaitra = (chiso2 - chiso1) * 8
    else:
        tienphaitra = (chiso2 - chiso1) * 10
    print(f"Ten chu ho: {hovatenchuho}")
    print(f"Tien dien phai tra: {tienphaitra}")
    #Tra ket qua
    return tienphaitra, hovatenchuho
tinhtiendien()
```

```python
# Bai 10 - Chu so hang tram, chuc, don vi
def chusohangtramchuchudonvi():
    # Nhap mot so nguyen co 3 chu so
    a = int(input("Nhap mot so nguyen co 3 chu so: "))
    # Tach ra hang tram, chuc, don vi
    hangtram = a // 100
    hangchuc = (a % 100) // 10
    hangdonvi = a % 10
    # In ra ket qua
    print("Chu so hang tram:", hangtram)
    print("Chu so hang chuc:", hangchuc)
    print("Chu so hang don vi:", hangdonvi)
    return hangtram, hangchuc, hangdonvi
chusohangtramchuchudonvi()
```

```python
# Bai 11 - Tinh dien tich va chu vi hinh vuong
def hinhvuong():
    # Nhap canh cua hinh vuong
    a = float(input("Nhap canh cua hinh vuong: "))
    # Tinh dien tich va chu vi
    dientich = a * a
    chuvi = 4 * a
    # In ra ket qua
    print("Dien tich hinh vuong:", dientich)
    print("Chu vi hinh vuong:", chuvi)
    return dientich, chuvi
hinhvuong()
```

```python
# Bai 12 - Tinh dien tich va chu vi hinh chu nhat
def hinhchunhat():
    # Nhap chieu dai va chieu rong
    Chieudai = float(input("Nhap chieu dai hinh chu nhat: "))
    Chieurong = float(input("Nhap chieu rong hinh chu nhat: "))
    # Tinh dien tich va chu vi
    dientich = Chieudai * Chieurong
    chuvi = 2 * (Chieudai + Chieurong)
    # In ra ket qua
    print("Dien tich hinh chu nhat:", dientich)
    print("Chu vi hinh chu nhat:", chuvi)
    return dientich, chuvi
hinhchunhat()
```

```python
# Bai 13 - Tinh dien tich và chu vi hinh tron
def hinhtron():
    # Nhap ban kinh hinh tron
    r = float(input("Nhap ban kinh hinh tron: "))
    # Tinh dien tich va chu vi
    dientich = 3.14 * r * r
    chuvi = 2 * 3.14 * r
    # In ra ket qua
    print("Dien tich hinh tron:", dientich)
    print("Chu vi hinh tron:", chuvi)
    return dientich, chuvi
hinhtron()
```

```python
# Bai 14 - Tam giác là tác giam
def tamgiac():
    a, b, c = map(float, input("Nhap 3 canh cua tam giac: ").split())
    # Kiem tra xem 3 canh co tao thanh tam giac khong
    if a + b > c and a + c > b and b + c > a:
        print("3 canh da nhap co tao thanh tam giac")
        # Tinh dien tich tam giac theo cong thuc Heron
        p = (a + b + c) / 2
        S = (p * (p - a) * (p - b) * (p - c)) ** 0.5
        print("Dien tich tam giac la:", S)
        return S
    else:
        print("3 canh da nhap khong tao thanh tam giac")
        return a, b, c
tamgiac()
```

```python
# Bai 15 - Tam giac deu
def tamgiacdeu():
    a, b, c = map(float, input("Nhap 3 canh cua tam giac: ").split())
    # Kiem tra xem 3 canh co tao thanh tam giac deu khong
    if a == b == c:
        print("Đây là tam giác đều")
    else:
        print("Đây không phải là tam giác đều")
    return a, b, c
tamgiacdeu()
```

```python
# Bai 17 - Tam giac vuong
def tamgiacvuong():
    a, b, c = map(float, input("Nhap 3 canh cua tam giac: ").split())
    # Kiem tra xem 3 canh co tao thanh tam giac vuong khong
    if a ** 2 + b ** 2 == c ** 2 or a ** 2 + c ** 2 == b ** 2 or b ** 2 + c ** 2 == a ** 2:
        print("Đây là tam giác vuông")
    else:
        print("Đây không phải là tam giác vuông")
    return a, b, c
tamgiacvuong()
```

```python
# Bai 18 - Phuong trinh bac 1
def phuongtrinhbac1():
    # Nhap a va b
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    # Giai phuong trình
    if a == 0:
        if b == 0:
            print("Phuong trinh co vo so nghiem")
        else:
            print("Phuong trinh vo nghiem")
    else:
        x = -b / a
        print(f"Nghiem duy nhat cua phuong trinh la: {x}")
    return a, b
phuongtrinhbac1()
```

```python
# Bai 19 - Bat phuong trinh bac 1
def batphuongtrinhbac1():
    # Nhap a va b
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    # Giai bat phuong trinh
    if a > 0:
        print(f"Bat phuong trinh co nghiem: x >= {-b / a}")
    elif a < 0:
        print(f"Bat phuong trinh co nghiem: x <= {-b / a}")
    else:
        if b >= 0:
            print("Bat phuong trinh co vo so nghiem")
        else:
            print("Bat phuong trinh vo nghiem")
    return a, b
batphuongtrinhbac1()
```

```python
# Bai 20 - Phuong trinh bac 2
def phuongtrinhbac2():
    # Nhap a, b, c
    a = float(input("Nhap he so a: "))
    b = float(input("Nhap he so b: "))
    c = float(input("Nhap he so c: "))
    # Giai phuong trinh bac 2
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phuong trinh co nghiem kep: {x}")
    else:
        x1 = (-b + delta ** 0.5) / (2 * a)
        x2 = (-b - delta ** 0.5) / (2 * a)
        print(f"Phuong trinh co hai nghiem phan biet: x1 = {x1}, x2 = {x2}")
    return delta
phuongtrinhbac2()
```

```python
# Bai 28 - May tinh 2 ngoi
def mayTinh2Ngoi():
    # Nhap hai so nguyen
    a = int(input("Nhap so nguyen thu nhat: "))
    b = int(input("Nhap so nguyen thu hai: "))
    # Nhap phep toan
    pheptoan = input("Nhap phep toan: ")
    # Thuc hien phep toan
    if pheptoan == "+":
        ketqua = a + b
        print(f"Tổng hai số là = {ketqua}")
    elif pheptoan == "-":
        ketqua = a - b
        print(f"Hiệu hai số là = {ketqua}")
    elif pheptoan == "*":
        ketqua = a * b
        print(f"Tích hai số là = {ketqua}")
    elif pheptoan == "/":
        if b != 0:
            ketqua = a / b
            print(f"Thương hai số là = {ketqua:.2f}")
        else:
            print("Khong the chia cho 0")
    else:
        print("Phep toan khong hop le")
    return a, b, pheptoan
mayTinh2Ngoi()
```

```python
# Bai 30 - Doi sang gio phut giay
def quydoiTime():
    # Nhap giay
    s = int(input("Nhap so giay: "))
    # Quy doi
    s1 = s % 60
    h = (s % 3600) // 3600
    m = s // 60 
    # In kết quả
    print(f"{h}:{m}:{s1}")
    return h, m, s1
quydoiTime()
```
