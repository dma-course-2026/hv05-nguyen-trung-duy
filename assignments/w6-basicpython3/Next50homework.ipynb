```python
# Bai 14 - Tam giác là tác giam
def tamgiac():
    a, b, c = map(float, input("Nhap 3 canh cua tam giac: ").split())
    # Kiem tra xem 3 canh co tao thanh tam giac khong
    if a + b > c and a + c > b and b + c > a:
        print("3 canh da nhap co tao thanh tam giac")
        # Tinh dien tich tam giac theo cong thuc Heron
        p = (a + b + c) / 2
        S = (p * (p - a) * (p - b) * (p - c)) ** 0.5
        print("Dien tich tam giac la:", S)
        return S
    else:
        print("3 canh da nhap khong tao thanh tam giac")
        return a, b, c
tamgiac()
```

```python
# Bai 32 - He phuong trinh bac 1 hai an
def tinhtoanhephuongtrinh():
    a, b, m, c, d, n = map(int, input("Nhap a, b, m, c, d, n: ").split(","))
    # Crammer
    D = a * d - b * c
    Dx = m * d - b * n
    Dy = a * n - c * m
    if D == 0 and Dx == Dy == 0:
        print ("Phương trình có vô số nghiệm")
    elif D == 0 and Dx > 0 and Dy > 0:
        print ("phương trình vô nghiệm")
    else:
        x = Dx/D
        y = Dy/D
        print (f"Hệ phương trình có 1 cặp nghiệm là {x} và {y}")
    return x, y
tinhtoanhephuongtrinh() 
```

```python
# Bai 33 - Diem thuoc duong tron
def cirpoint():
    # Nhap tu ban phim
    r = int(input("Nhập bán kính: "))
    x0 = int(input("Nhập tọa độ tâm đường tròn x: "))
    y0 = int(input("Nhập tọa độ tâm đường tròn y: "))
    xa = int(input("Nhập tọa độ xa: "))
    ya = int(input("Nhập tọa độ ya: "))
    # Tinh khoang cach
    d = d = ((xa - x0)**2 + (ya - y0)**2)**0.5
    # Kiem tra xem điểm A có thuộc đường tròn hay không
    if d < r:
        print ("Điểm A nằm trong đường tròn")
    elif d == r:
        print ("Điểm A nằm trên đường tròn")
    else:
        print ("Điểm A nằm ngoài đường tròn")
    return r, d
cirpoint()    
```

```python
# Bài 40 - Tổng các số chia hết cho 3 hoặc 5 trong [M, N]
def tong():
    # Nhập vào 2 số nguyên dương
    M = int(input("Nhập vào số nguyên dương M: "))
    N = int(input("Nhập vào số nguyên dường N: "))
    tong = 0
    # Xử lý tính toán
    for i in range (M, N + 1):
        if M < N:
            if i % 3 == 0 or i % 5 == 0:
                tong += i
    if M >= N:
        print("Không thực hiện được phép toán vì M phải nhỏ hơn N")
    else:
        print(f"Tổng các số chia hết cho 3 hoặc 5 trong [{M}, {N}] = {tong}")
    return tong
tong()
```

```python
# Bai 44 - Đếm số trong khoảng (10, 20)
def countinRange():
    # Nhập N số
    N = int(input("Nhập số bất kỳ: "))
    # Xử lý đếm
    dem = 0
    for i in range (1, N + 1):
        so = int(input(f"Nhập số thứ {i}: "))
        if 10 < so < 20:
            dem += 1
    if dem == 0:
        print ("Không có số nào trong khoảng 10 đến 20")
    else:
        print (f"Số các số > 10 và < 20 là {dem}")
    return dem
countinRange()
```

```python
# Bai 45 - Kiểm tra số nguyên tố
def songuyento():
    # Nhập vào số nguyên
    n = int(input("Nhập vào số nguyên: "))
    # Kiểm tra n có phải số nguyên tố không
    if n < 2:
        print(n, "không phải là số nguyên tố")
        return
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print(n, "không phải là số nguyên tố")
            return
    print(n, "là số nguyên tố")
songuyento()
```

```python
# Bai 46 - Kiem tra số hoàn hảo
def perfectNumber():
    # Nhập vào số nguyên
    n = int(input("Nhập vào số nguyên: "))
    tong = 0
    # Kiểm tra n có phải số hoàn hảo không
    for i in range (1, n - 1):
        if n % i == 0:
            tong += i
    if tong == n:
        print (n, "là số hoàn hảo")
    else:
        print (n, "không là số hoàn hảo")
    return n, tong
perfectNumber()
```

```python
# Bai 47 - Các số hoàn hảo nhỏ hơn n
def perfectNumber():
    # Nhập n
    n = int(input("Nhập số n: "))
    perfectNumberList = []
    # Tìm số hoàn hảo nhỏ hơn n
    for i in range(1, n):
        tong = 0
        for i1 in range(1, i):
            if i % i1 == 0:
                tong += i1
        if tong == i:
            perfectNumberList.append(i)
    if perfectNumberList:
        print (f"Các số hoàn hảo nhỏ hơn {n} là {perfectNumberList}")
    else:
        print (f"Không có số hoàn hảo nào nhỏ hơn {n}")
    return tong
perfectNumber()
```

```python
# Bai 48 - Số chính phương
import math
def timSoChinhPhuong():
    # Input số n
    n = int(input("Nhập vào số nguyên dương n: "))
    m = math.sqrt(n)
    # Kiểm tra n có phải số chính phương không
    if m*m == n:
        print (n, "là số chính phương")
    else:
        print (n, "không là số chính phương")
    return m, n
timSoChinhPhuong()
```

```python
# Bai 49 - Số Armstrong
def armStrong():
    armStrongNumber = []
    # Duyệt để ra số có 3 chữ số
    for hangtram in range (1, 10):
        for hangchuc in range (0, 10):
            for hangdonvi in range (0, 10):
                so3chuso = hangtram * 100 + hangchuc * 10 + hangdonvi
                if so3chuso == hangtram ** 3 + hangchuc ** 3 + hangdonvi ** 3:
                    armStrongNumber.append(so3chuso)
    print (f"Số ArmStrong là: {armStrongNumber}")
armStrong()
```
