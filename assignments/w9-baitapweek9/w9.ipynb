```python
# Bài 1
def sumTwoNumber(a, b):
    return a + b
sum = sumTwoNumber(10, 5)
print (f"Tổng 2 số là: {sum}")
```

```python
# Bài 2
def toString(a):
    return str(a)
toStringA = toString(5)
print (f"Kết quả sau khi ép kiểu: {toStringA}")
```

```python
# Bài 3
def toIntForSum(a, b):
    return str(a), str(b)
a_str, b_str = toIntForSum(5, 10)
sum = int(a_str) + int(b_str)
print(f"Tổng 2 số nguyên là: {sum}")
```

```python
# Bài 4
def joinStr():
    stringA = input("Nhập vào chuỗi 1: ")
    stringB = input("Nhập vào chuỗi 2: ")
    to1String = stringA + " " + stringB
    print (f"Chuỗi sau khi ghép: {to1String}")
joinStr()
```

```python
# Bài 5
def lenAandB():
    stringA = input("Nhập vào chuỗi 1: ")
    stringB = input("Nhập vào chuỗi 2: ")
    if len(stringA) > len(stringB):
        print(stringA)
    elif len(stringA) == len(stringB):
        to1String = stringA + " " + stringB
        print (to1String)
    else:
        print (stringB)
lenAandB()
```

```python
# Bài 6
def isChanorLe():
    n = int(input("Nhập một số nguyên: "))
    if n % 2 == 0:
        print(f"{n} là số chẵn")
    else:
        print(f"{n} là số lẻ")
isChanorLe()
```
