```python
#Bai 7
a, b, c = map(float, input("Nhập 3 điểm: ").split())
diem_trung_binh = (a + b + c) / 3
print(f"Điểm trung bình là: {diem_trung_binh}")
if diem_trung_binh >= 5.0:
    print("Đậu")
else:
    print("Rớt")
```

```python
#Bai 8
ch = input("Nhập một ký tự: ")
if len(ch) == 1 and ("a" <= ch <= "z" or "A" <= ch <= "Z"):
    print(f"{ch} là ký tự Alphabet")
else:
    print(f"{ch} không phải là ký tự Alphabet")
```

```python
#Bai 9
hotenchuho = input("Nhập họ tên chủ hộ: ")
chisodienthangtruoc = int(input("Nhập chỉ số điện tháng trước: "))
chisodienthang_nay = int(input("Nhập chỉ số điện tháng này: "))
luongdientieuthu = chisodienthang_nay - chisodienthangtruoc
if luongdientieuthu <= 60:
    tiendienphaitra = luongdientieuthu * 5
elif luongdientieuthu <= 160:
    tiendienphaitra = 60 * 5 + (luongdientieuthu - 60) * 8
else:
    tiendienphaitra = 60 * 5 + 100 * 8 + (luongdientieuthu - 160) * 10
print(f"Chủ hộ: {hotenchuho}")
print(f"Số tiền phải trả là: {tiendienphaitra:.2f}")
```
