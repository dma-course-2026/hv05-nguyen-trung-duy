```python
#Bai 6
a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))

dien_tich_con_lai = a * b - 3.14 * (a / 2) ** 2

print(f"Diện tích còn lại: {dien_tich_con_lai:.2f}")
```

```python
#Bai 7
a, b, c = map(float, input("Nhập 3 điểm: ").split())

diem_trung_binh = (a + b + c) / 3
print(f"Điểm trung bình là: {diem_trung_binh}")

if diem_trung_binh >= 5.0:
    print("Đậu")
else:
    print("Rớt")
```

```python
#Bai 8
ch = input("Nhập một ký tự: ")

if len(ch) == 1 and ("a" <= ch <= "z" or "A" <= ch <= "Z"):
    print(f"{ch} là ký tự Alphabet")
else:
    print(f"{ch} không phải là ký tự Alphabet")
```

```python
#Bai 9
ho_ten = input("Nhập họ tên chủ hộ: ")
chi_so_thang_truoc = int(input("Nhập chỉ số điện tháng trước: "))
chi_so_thang_nay = int(input("Nhập chỉ số điện tháng này: "))

luong_dien = chi_so_thang_nay - chi_so_thang_truoc

if luong_dien <= 60:
    tien_dien = luong_dien * 5
elif luong_dien <= 160:
    tien_dien = 60 * 5 + (luong_dien - 60) * 8
else:
    tien_dien = 60 * 5 + 100 * 8 + (luong_dien - 160) * 10

print(f"Chủ hộ: {ho_ten}")
print(f"Số tiền phải trả là: {tien_dien:.2f}")
```

```python
#Bai 10
n = int(input("Nhập số nguyên có 3 chữ số: "))

hang_tram = n // 100
hang_chuc = (n % 100) // 10
hang_don_vi = n % 10

print(f"Hàng trăm = {hang_tram}")
print(f"Hàng chục = {hang_chuc}")
print(f"Hàng đơn vị = {hang_don_vi}")
```
