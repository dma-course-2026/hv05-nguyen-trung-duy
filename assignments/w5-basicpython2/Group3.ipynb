```python
# Bai 38
n = int(input("Nhập n: "))
if n < 0:
    print("Giai thừa không xác định với số nguyên âm.")
else:
    giai_thua = 1
    for i in range(1, n + 1):
        giai_thua = giai_thua * i
    print(f"{n} giai thừa = {giai_thua}")
```

```python
# Bai 39
a = int(input("Nhập cơ số a: "))
n = int(input("Nhập số mũ n: "))
res = 1
for i in range(n):
    res = res * a
luythua = res
print(f"{a} lũy thừa {n} = {luythua}")
```

```python
# Bai 40
M = int(input("Nhập M: "))
N = int(input("Nhập N: "))
tong = 0
for i in range(M, N + 1):
    if i % 3 == 0 or i % 5 == 0:
        tong = tong + i
print(f"Kết quả: {tong}")
```
