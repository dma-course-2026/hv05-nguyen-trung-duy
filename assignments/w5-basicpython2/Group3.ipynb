```python
# Bai 35 - Tính tổng từ 1 đến n
n = int(input("Nhập n: "))
tong = 0

for i in range(1, n + 1):
    tong += i

print(f"Tổng từ 1 đến {n} = {tong}")
```

```python
# Bai 38 - Tính giai thừa n!
n = int(input("Nhập n: "))

if n < 0:
    print("Giai thừa không xác định với số nguyên âm.")
else:
    giai_thua = 1
    for i in range(1, n + 1):
        giai_thua *= i
    print(f"{n} giai thừa = {giai_thua}")
```

```python
# Bai 39 - Tính lũy thừa a^n
a = float(input("Nhập cơ số a: "))
n = int(input("Nhập số mũ n: "))

ket_qua = a ** n
print(f"{a:g} lũy thừa {n} = {ket_qua:g}")
```

```python
# Bai 40 - Tổng các số chia hết cho 3 hoặc 5 trong [M, N]
M = int(input("Nhập M: "))
N = int(input("Nhập N: "))
tong = 0

for i in range(M, N + 1):
    if i % 3 == 0 or i % 5 == 0:
        tong += i

print(f"Kết quả: {tong}")
```
