```python
#Bai 14
import math

a, b, c = map(float, input("Nhập 3 cạnh a, b, c: ").split())

if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
    chu_vi = a + b + c
    p = chu_vi / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))

    print(f"Chu vi: {chu_vi:.2f}")
    print(f"Diện tích: {dien_tich:.2f}")
else:
    print("Ba cạnh không tạo thành tam giác.")
```

```python
#Bai 20
import math

a, b, c = map(float, input("Nhập a, b, c: ").split())

if a == 0:
    if b == 0:
        if c == 0:
            print("Phương trình có vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print(f"Phương trình bậc nhất có nghiệm: x = {-c / b}")
else:
    delta = b ** 2 - 4 * a * c

    if delta < 0:
        print("Phương trình vô nghiệm trong tập số thực")
    elif delta == 0:
        print(f"Phương trình có nghiệm kép: x = {-b / (2 * a)}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print(f"x1 = {x1}, x2 = {x2}")
```

```python
#Bai 32

a, b, m, c, d, n = map(float, input("Nhập a, b, m, c, d, n: ").split())

D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"x = {x:.3f}, y = {y:.3f}")
elif Dx == 0 and Dy == 0:
    print("Hệ phương trình có vô số nghiệm")
else:
    print("Hệ phương trình vô nghiệm")
```

```python
#Bai 33
import math

x0, y0, r = map(float, input("Nhập tâm x0, y0 và bán kính R: ").split())
xa, ya = map(float, input("Nhập tọa độ điểm A (xa, ya): ").split())

if r < 0:
    print("Bán kính không hợp lệ")
else:
    khoang_cach = math.sqrt((xa - x0) ** 2 + (ya - y0) ** 2)

    if math.isclose(khoang_cach, r):
        print("Điểm A nằm trên đường tròn")
    elif khoang_cach < r:
        print("Điểm A nằm trong đường tròn")
    else:
        print("Điểm A nằm ngoài đường tròn")
```
